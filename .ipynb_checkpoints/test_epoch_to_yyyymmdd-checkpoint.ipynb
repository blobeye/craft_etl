```python
year0 = "2019"
year1 = str(int(year0) + 1)

datelist = [
    year0 + "0701",
    year0 + "0801",
]

i = 0;
dates = {"start_date": datelist[i], "end_date": datelist[i+1]}
```

```python
# #Function to convert Unix timestamp (milliseconds from 1970) to YYYYMMDD string

# @F.udf("string")
# def to_date(ts):
#     return datetime.utcfromtimestamp(ts / 1000).strftime("%Y%m%d")
```

```python
print("   retrieving flightplans: ", datetime.now())

df_flightplanseries = (
    api.dataframe("FlightplanSeries", **dates, metadata=True)
    .select(
        "track_key",
#        to_date("metadata.effective_start_date").alias("start_date"),
#        to_date("metadata.effective_end_date").alias("end_date"),
        "metadata.effective_start_date",
        "metadata.effective_end_date",
    )
)
```

```python
(
    df_flightplanseries
    .withColumn('start_date',F.from_unixtime(F.col('effective_start_date')/1000, 'yyyyMMdd'))
    .withColumn('end_date',F.from_unixtime(F.col('effective_end_date')/1000, 'yyyyMMdd'))
    .show()
)
```

```python
print("   completed flightplans: ", datetime.now())
```
